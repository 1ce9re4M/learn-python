```python
# 練習問題 -------------------------------------------------------------
def var(lst):
    # 分散の求め方
    # (平均ー各データ)**2 / 要素数
    
    # lst = [5,10,15]
    # a = sum(lst)
    # ave = a / len(lst)
    # var = 0
    # for x in lst:
    #     b = (x - ave) ** 2 / len(lst)
    #     var += b
    # print(var)

    return sum([((x - (sum(lst)/len(lst))) ** 2 / len(lst)) for x in lst])

# 模範解答 -------------------------------------------------------
def var_T(lst):
    n = len(lst)
    av = sum(lst)/n
    return sum([(x - av)*(x - av) for x in lst])

# print(var([3,4,1,2]))
print(var([3,4,1,2]) == 1.25)
```

```python
# 内包表記の入れ子
[[x*y for y in range(x+1)] for x in range(4)]
# [[0], [0, 1], [0, 2, 4], [0, 3, 6, 9]]

# 1次元のリストになる
[x*y for x in range(4) for y in range(x+1)]
# [0, 0, 1, 0, 2, 4, 0, 3, 6, 9]

def allsubstrings(s):
    return [s[i:j] for i in range(len(s)) for j in range(i+1,len(s)+1)]
allsubstrings("abc")
# ["a", "ab", "abc", "b", "bc", "c"]


def allsubstrings_nor(s):
    result = []
    for i in range(len(s)):
        print("i：",i)
        for j in range(i+1, len(s)+1):
            print("j：",j)
            print(s[i:j])
            print("")
            result.append(s[i:j])
    return(result)
print(allsubstrings_nor("abc"))
```

```python
# 練習問題 ------------------------------------------------------------
# list2 = [[1,2], [3], [1,2,3]]
# total_list2 = 12
# total = 0
# for i in list2:
#     print(i)
#     for j in i:
#         total += j
# print(total)

def sum_lists(list1):
    num_list =  [j for i in list1 for j in i]
    return sum(num_list)

# 模範解答 ---------------------------------------------------------
def sum_lists(list1):
    return sum([j for i in list1 for j in i])



print(sum_lists([[20, 5], [6, 16, 14, 5], [16, 8, 16, 17, 14], [1], [5, 3, 5, 7]]) == 158)
```

```python
def sum_matrix(list1, list2):
    return [[list1[i][j] + list2[i][j] for j in range(3)] for i in range(3)]

print(sum_matrix([[1,2,3], [1,2,3], [1,2,3]],[[4,5,6], [4,5,6], [4,5,6]]))
print(sum_matrix([[1,5,3],[4,5,6],[7,8,9]], [[1,4,7],[2,5,8],[3,6,9]])==[[2, 9, 10], [6, 10, 14], [10, 14, 18]])
```

```python
# 条件付き内包表記 -------------------------------------------------
words = ["cat", "dog", "elephant", None, "giraffe"]
length= [len(w) for w in words if w != None]
print(length) # [3, 3, 8, 7]

# セット内包表記 -----------------------------------------------------
words = ["apple", "orange", "banana", "pine", "grape"]
length_set = {len(w) for w in words}
print(length_set) # {4, 5, 6}
# set()
# 空のセットを作りたいときに使う

# 辞書内包表記 --------------------------------------------------------
words = ["one", "two", "three", "four", "five"]
length_dic = {w:len(w) for w in words}
print(length_dic)
# {'one': 3, 'two': 3, 'three': 5, 'four': 4, 'five': 4}
length_rdic = {len(w):w for w in words}
print(length_rdic)
```
