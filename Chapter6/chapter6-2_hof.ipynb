```python
# 高階層関数 higher-order function
# 関数を引数として受け取ったり返したりするもの


# max
ls = [3, 0, 18, -5, 9, -4, 20, -30]
max(ls)
# キーワード引数を設定できる
# abs -> 絶対値
print(max(ls, key = abs))
print("")

# キーワード引数
# 条件を設定したうえで該当のものを取り出してくれる
print(min("Apple", "banana", key = str.lower))
```

```python
# sorted()-----------------------------------------------------
print(sorted(ls, key = abs))

# 昇順に並べ替える
def invert(x):
    return -x
print("\n"+str(sorted(ls, key=invert)))
# reverseもある(リストのみ)
print(sorted(ls, reverse = True))
```

```python
# ラムダ式 ----------------------------------------------------------
# 上記のdef invert(x): のような簡単な定義を簡単に書ける
sorted(ls, key = lambda x: -x)
```

```python
# リストからイテラブルへ ------------------------------------------
a = (3, -8, 1, 0, 7, -5)
b = ("hello world")

print(max(a))
print(sorted(a))
print(max(b))
print(sorted(b))

c = {10:10, 5:2, 9:1}
print(max(c))
print(sorted(c))
```
