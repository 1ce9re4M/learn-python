```python
import numpy as np
a1 = np.array([0, 1, 2, 3, 4, 5])
a2 = a1.reshape(2,3)    # (2,3)行列
print("\n(1)-1 reshape()")
print(a2)

a1[1] = 6
# a1 a2 は内部的にデータを共有している
print("\n(1)-2")
print(a1)
print(a2)

# ravel()-------------------------------------------------------
# 多次元配列を1次元に戻せる
a = np.array([0,1,2,3,4,5]).reshape(2,3)
print("\n(2)-1 ravel()")
print(a.ravel())

elems = np.array([0,1,2,3,4,5])
a = elems.reshape(2,3).ravel()
elems[1] = 6
print("(2)-2")
print(a)
```

```python
# arange(), linspace(), zeros, ones

# arange関数--------------------------------------------------------
# array rangeの略
# 開始値、終了値、刻み幅を指定して配列を生成　リストの扱いと同じかんじ
print("(1) arange関数")
print(np.arange(3))
print(np.arange(0, 1, 0.2))

# linspace関数---------------------------------------------------------
# 範囲を等分割した配列を生成
print("\n(2) linspace関数")
print(np.linspace(0,10,5))

# zeros と ones-------------------------------------------------------
# zeros : 0からなる配列を生成
# ones : 1からなる配列を生成
print("\n(3) zerosとones")
print("zeros")
print(np.zeros(4))
print(np.zeros((2,3)))
print("ones")
print(np.ones(4))
print(np.ones((2,3)))
```

```python
# 練習問題-----------------------------------------------------------------
print("\n練習問題")
def arange_square_matrix(n):
    result = []
    for i in range(n):
        a = np.arange(i, i+n)
        result.append(a)
    # print(result)　
    #[array([0, 1, 2]), array([1, 2, 3]), array([2, 3, 4])]
    # ただのリスト状態
    return np.array(result) #array()配列化

# 模範解答 -----------------------------------------------------------
def arange_square_matrix_T(n):
    return np.arange(n) + np.arange(n)[:,None]
# np.arange -> n行と指定した
# [:,None] (.reshape(-1,1)も同様) 
#    ➥ n行のものを縦1列にする
# つまり。。。（ｎ, １）＋（１, ｎ）＝（ｎ, ｎ）ということ！！

print(arange_square_matrix(3))
print(all(map(all,(arange_square_matrix_T(3) == np.array([[0,1,2],[1,2,3],[2,3,4]])))))
```

```python
# 配列要素の操作

# インデックスアクセス
print("インデックスアクセス")
a = np.arange(3)
print(a)
print(a[1])
a[-1] = 3
print(a)

a = np.arange(6).reshape(2,3)
print(a)

# 行、列をまとめて指定
# print(a[1,2])   # 5
a[1,2] = 6
print(a)

# スライス
print("\nスライス")
a = np.arange(5)    # [0 1 2 3 4]
print(a[1:4])   # [1 2 3]
print(a[1:])    # [1 2 3 4]
print(a[:-2])   # [0 1 2]
print(a[::2])   # [0 2 4]
print(a[::-1])  # [4 3 2 1 0]
```

```python
a = np.arange(5)
print(a)    # [0 1 2 3 4]
a[1:4] = 6
print(a)    # [0 6 6 6 4]
a = np.arange(5)
a[::2] = 6
print(a)    # [6 1 6 3 6]

# リストでは不可
a = [0, 1, 2, 3, 4]
a[1:4] = [6]    # [0, 6, 4]

a = np.arange(9).reshape(3,3)
# [[0 1 2]
#  [3 4 5]
#  [6 7 8]]

print(a[:2, :2])
# [[0 1]
#  [3 4]]

print(a[1:,1:])
# [[4 5]
#  [7 8]]
```

```python
# for文------------------------------------------------------------
print("(1)")
for v in np.arange(3):
    print(v)
    # 0 1 2縦書き

print("\n(2)")
for row in np.arange(6).reshape(2,3):
    print(row)
    # [0 1 2]
    # [3 4 5]

print("\n(3)")
# np.ndenumerate() ----------------------------------------------------
# 多次元配列版のenumerate関数（番号の割り振り）
for idx, e in np.ndenumerate(np.arange(6).reshape(2,3)):
    print(idx, e)
    # (0, 0) 0
    # (0, 1) 1
    # (0, 2) 2 ....

print("\n(4)")
for idx, e in np.ndenumerate(np.arange(3)):
    print(idx, e)

# arange()は生み出す、array()は既存のものを配列にする！！
```

```python
# 配列のスカラ演算　＊＊ 超大事！！！＊＊
a = np.arange(4)    # [0 1 2 3]

print(a + 1) # 各要素に１を加算
print(a ** 2) # 各要素を2乗

b = a+1 # [1 2 3 4]
print(1 / b) # [1. 0.5 0.333 0.25]
print( 9 // b) # [9 4 3 2]

print("\n配列同士の計算")
# 線形での計算とは別
a = np.arange(4).reshape(2,2) # [[0 1] [2 3]]
b = np.arange(1,5).reshape(2,2) # [[1 2] [3 4]]

print(a + b) # [[1 3] [5 7]]
print(a * b) # [[0 2] [6 12]]
print(a / b) # [0 0.5] [0.666 0.75]]

print("")
c = 3 * a # [[0 3] [6 9]]
print(c // b) # [[0 1] [ 2 2]]
print(a ** b) # [[0 1] [8 81]]
```

```python
# ユニバーサル関数
# sqrt,log などの特定の演算を与えた値を返す関数のこと
a = np.zeros(3) + 2 # [2. 2. 2.]
print(np.sqrt(a)) # [1.41421356  1.41421356  1.41421356]
b = np.zeros((2,2)) + 2 # [[2 2] [2 2]]
print(np.sqrt(2)) # [1.41421356]  0次元配列を扱うことも可能
```

```python
# dot()  線形での積と同様！！！！
print("dot関数")
np.dot(np.arange(4), np.arange(1,5))
print(np.dot(np.arange(4).reshape(2,2), np.arange(1,5).reshape(2,2)))
# [[0 1]  *  [[1 2]   =  [[3  4]]
#  [2 3]]     [3 4]]      [11 16]

# sort()
print("\nsort関数")
a = np.array([3, 4, -1, 0, 2])
a.sort() # インプレースなのでsort()事態はNoneを返す＝print(a.sort())はだめ
print(a) # [-1, 0, 2, 3, 4]

# sum, max, min, mean 関数
print("\nsum, max, min, mean 関数")
a = np.arange(6).reshape(2,3) # [[0 1 2] [3 4 5]]
print(a.sum()) # 15
print(a.sum(0)) # [3 5 7]
```

```python
# 配列の保存と復元　savetxt(), loadtxt()

# savetxt() -------------------------------------------------------
# ファイル名を付けテキストで保存
np.savetxt("arange3.txt", np.arange(3))
np.savetxt("arange2x3.txt", np.arange(6).reshape(2,3))

# loadtxt() ----------------------------------------------------------
# ファイルを出力する
a = np.loadtxt("arange2x3.txt")
print(a)
print("")
print(np.loadtxt("arange3.txt"))

# .txtと.gz -------------------------------------------------------
# .gzにするとzipファイルとして保存、または解凍して復元する
```

```python
# 真理値配列
# 条件を付けることでTrue、Falseで返す
# インデックス指定をすることで特定のインデックスを返してくれる
a = np.arange(6).reshape(2,3)
print("True Falseで返す")
print(a) # [[0 1 2] [3 4 5]]
print(a < 3) # [[True True True] [False False False]]

print("\nインデックスとして利用")
a = np.array([0, 1, 2, -3, -4, 5, -6, -7])
print(a[a < 0]) # [-3 -4 -6 -7]
print(a[(a < 0) & (a % 2 == 0)]) # [-4 -6]
a[a<0] = 5 # [0 1 2 5 5 5 5]
print(a)
```

```python
# 線形代数の演算 matmul() ---------------------------------------------------
# matmul() -> 行列積専用の関数 .dot()と近縁
# .identity() 単位行列を作れちゃうよ
I = np.identity(3)
print(I) # [[1 0 0] [0 1 0] [0 0 1]]
a = np.arange(9).reshape(3,3) # [[0 1 2] [3 4 5] [6 7 8]]
print(np.matmul(a, I)) # [[0 1 2] [3 4 5] [6 7 8]]

# .linalg.norm() -> ノルム(ベクトルの長さ、大きさ)を求める
# 例：(0,0)から(3,4)までのノルム
# √(3**2 + 4**2) = √25 = 5
print("\nノルムを求める")
print(str(np.ones(3)) + "のノルム")
print(np.linalg.norm(np.ones(3)))
```
